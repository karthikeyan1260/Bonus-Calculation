<b> Employee Bonus Calculation </b>

```python
def calculate_bonus (salary, rating):
    rating = rating.lower()

    if rating == "excellent":
        bonus = salary * 0.20
    elif rating == "good":
        bonus = salary * 0.10
    elif rating == "average":
        bonus = salary * 0.05
    elif rating == "poor":
        bonus = salary * 0.0
    else:
        return none

    return bonus

def display_report (name, salary, rating, bonus):
    annual_salary = (salary * 12) + bonus

    print("\n----- Employee Report -----")
    print("Employee Name:", name)
    print("Monthly Salary:", salary)
    print("Performance Rating:", rating.title())
    print("Monthly Bonus:", bonus)
    print("Annual Salary with Bonus:", annual_salary)

# Get employee details
name = input("Enter Employee Name: ")
salary = float(input("Enter Monthly Salary: "))
rating = input("Enter Performance Rating: ")

#Calculate Bonus
bonus = calculate_bonus (salary, rating)

# Validate rating and display report
if bonus is None:
    print("Error: Invalid performance rating!")
    print("Please enter Excellent, Good, Average, or Poor.")
else:
    display_report(name, salary, rating, bonus)
        
```
